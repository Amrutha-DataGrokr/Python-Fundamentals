## 1. Creating Sets

A set is a mutable collection of distinct hashable objects. Sets do not support sequence-style indexing.

**Official:** https://docs.python.org/3/library/stdtypes.html#set-types-set-frozenset

```python
empty = set()
numbers = {1, 2, 3}
from_list = set([1, 2, 2, 3])
from_range = set(range(4))

print(empty) # An empty set
print(numbers) # A set with three unique numbers
print(from_list) # A set created from a list, duplicates are removed
print(from_range) # A set created from a range of numbers, includes 0, 1, 2, 3

print(type({}))  # {} is an empty dictionary
```

## 2. Membership and Iteration

Set membership is a key use case. Sets are iterable, but they do not provide positional indexing.

```python
s = {10, 20, 30}

print(20 in s) # True, because 20 is an element of the set s
print(99 not in s) # True, because 99 is not an element of the set s

for item in s:
    print(item)

try:
    print(s[0]) 
except TypeError as e:
    print(type(e).__name__, e) # TypeError: 'set' object is not subscriptable
```

## 3. `add()`

Adds an element to the set. The set is modified in place.

```python
s = {1, 2}
result = s.add(3) # Returns None, because add() modifies the set in place and does not return a new set.

print(s)
print(result)
```

## 4. `remove()` and `discard()`

`remove(x)` removes `x` and raises `KeyError` if absent. `discard(x)` removes `x` if present and does nothing if absent.

```python
s = {1, 2, 3}

s.remove(2) # Removes the element 2 from the set s. If 2 is not present, it raises a KeyError.
print(s) 

s.discard(99) # Removes the element 99 from the set s. If 99 is not present, it does nothing.
print(s)

try:
    s.remove(99)
except KeyError as e:
    print(type(e).__name__, e)
```

## 5. `pop()`

Removes and returns an arbitrary element. The order of removal should not be relied upon.

```python
s = {10, 20, 30}
item = s.pop() # Removes and returns an arbitrary element from the set s. 
#If the set is empty, it raises a KeyError.

print("removed:", item)
print("remaining:", s)
```

## 6. `clear()` and `copy()`

`clear()` removes all elements. `copy()` returns a shallow copy of the set.

```python
s = {1, 2, 3}
copy_s = s.copy() # Creates a shallow copy of the set s.

print(s is copy_s) # False, because s and copy_s are two different set objects.
print(s == copy_s) # True, because s and copy_s have the same elements.

result = s.clear() # Removes all elements from the set s. The set becomes empty.
print(s)
print(result)
```

## 7. `union()` and `update()`

`union()` returns a new set containing elements from the operands. `update()` adds elements from the operands to the existing set.

```python
a = {1, 2, 3}
b = {3, 4, 5}

print(a.union(b)) # Returns a new set containing all unique elements from both sets a and b.
print(a | b) # | is the union operator for sets, equivalent to a.union(b).

c = a.copy() # Creates a shallow copy of the set a.
result = c.update(b) # Updates the set c with all unique elements from set b.
print(c)
print(result)
```

## 8. `intersection()` and `intersection_update()`

`intersection()` returns common elements in a new set. `intersection_update()` keeps only common elements in the existing set.

```python
a = {1, 2, 3}
b = {2, 3, 4}

print(a.intersection(b)) # Returns a new set containing only the elements that are present in both sets a and b.
print(a & b) # & is the intersection operator for sets, equivalent to a.intersection(b).

c = a.copy() # Creates a shallow copy of the set a.
result = c.intersection_update(b) # Updates the set c to keep only the elements that are present in both sets c and b.
print(c)
print(result)
```

## 9. `difference()` and `difference_update()`

`difference()` returns elements present in the first set but not the others. `difference_update()` removes those elements from the existing set.

```python
a = {1, 2, 3, 4}
b = {3, 4, 5}

print(a.difference(b)) # Returns a new set containing elements that are in set a but not in set b.
print(a - b) # - is the difference operator for sets, equivalent to a.difference(b).

c = a.copy() # Creates a shallow copy of the set a.
result = c.difference_update(b) # Updates the set c to remove elements that are present in set b.
print(c)
print(result)
```

## 10. `symmetric_difference()` and `symmetric_difference_update()`

Symmetric difference contains elements in either set, but not in both.

```python
a = {1, 2, 3}
b = {3, 4}

print(a.symmetric_difference(b)) # Returns a new set containing elements that are in either set a or set b but not in both.
print(a ^ b) # ^ is the symmetric difference operator for sets, equivalent to a.symmetric_difference(b).

c = a.copy() # Creates a shallow copy of the set a.
result = c.symmetric_difference_update(b) # Updates the set c to keep only elements that are in either set c or set b but not in both.
print(c)
print(result)
```

## 11. `isdisjoint()`, `issubset()`, `issuperset()`

These methods test relationships between sets.

```python
a = {1, 2, 3}
b = {2, 3}
c = {4, 5}

print(a.isdisjoint(c)) # True, because sets a and c have no elements in common.
print(b.issubset(a)) # True, because all elements of set b are in set a.
print(a.issuperset(b)) # True, because all elements of set b are in set a.

print(b <= a) # True, because b is a subset of a.
print(a >= b) # True, because a is a superset of b.
print(b < a) # True, because b is a proper subset of a (b is a subset of a and b is not equal to a).
print(a > b) # True, because a is a proper superset of b (a is a superset of b and a is not equal to b).
```

## 12. Set Operators

| Operator | Meaning |
|---|---|
| `A | B` | Union |
| `A & B` | Intersection |
| `A - B` | Difference |
| `A ^ B` | Symmetric difference |
| `A <= B` | Subset |
| `A < B` | Proper subset |
| `A >= B` | Superset |
| `A > B` | Proper superset |

```python
A = {1, 2, 3}
B = {3, 4}

print(A | B)
print(A & B)
print(A - B)
print(A ^ B)
```

## 13. `frozenset`

`frozenset` is the immutable set type. It can be used where a hashable set-like object is required.

```python
fs = frozenset([1, 2, 3]) 
# Creates an immutable set, which means that its elements cannot be changed after creation.

print(fs)
print(type(fs))
print(2 in fs) # True, because 2 is an element of the frozenset fs

d = {fs: "value"} 
# Creates a dictionary with the frozenset fs as a key and "value" as its corresponding value.
print(d)

try:
    fs.add(4)
except AttributeError as e:
    print(type(e).__name__, e)
```

## 14. Complexity Quick Reference

| Operation | Typical complexity |
|---|---:|
| Membership | O(1) average |
| `add()` | O(1) average |
| `remove()` | O(1) average |
| `discard()` | O(1) average |
| `pop()` | O(1) average |
| `copy()` | O(n) |
| Union | O(n + m) |
| Intersection | Depends on operand sizes |
| Difference | Depends on operand sizes |
| Symmetric difference | Depends on operand sizes |

Set operations are hash-table based in CPython. Average-case complexity is not the same as a universal language guarantee.

## 15. Memory / Storage Notes

- Set elements must be hashable.
- Sets maintain uniqueness.
- Sets are designed around efficient membership testing.
- Sets do not provide sequence indexing.
- `copy()` creates a new set containing references to the same element objects.
- Avoid relying on set iteration order.
