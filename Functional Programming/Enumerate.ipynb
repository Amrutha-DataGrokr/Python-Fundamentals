# enumerate()

## 1. Definition:

`enumerate(iterable, start=0)` returns an enumerate iterator.

Each iteration produces a tuple:

```text
(index, value)
```

The index starts at `0` by default or at the value supplied to `start`.


````python
items = ["apple", "banana", "cherry"]

result = enumerate(items)

print(result)
print(list(result))
````

## 2. Basic iteration

The most common use is unpacking the index and value directly in a `for` loop.


````python
items = ["apple", "banana", "cherry"]

for index, item in enumerate(items):
    print(index, item)
````

## 3. Custom starting index

Use `start=` when the displayed numbering should begin elsewhere.

This is especially common for human-facing numbering, where counting starts at 1.


````python
items = ["apple", "banana", "cherry"]

for number, item in enumerate(items, start=1):
    print(f"{number}. {item}")
````

## 4. Why not `range(len(items))`?

This works:

```python
for i in range(len(items)):
    print(i, items[i])
```

But when we need both the index and the value, `enumerate()` communicates the intent directly and avoids indexing into the sequence.


````python
items = ["A", "B", "C"]

# Less direct when both index and value are needed:
for i in range(len(items)):
    print(i, items[i])

# Idiomatic:
for i, item in enumerate(items):
    print(i, item)
````

## 5. `enumerate()` works with any iterable

It does not require a list. It works with strings, tuples, generators, files, and other iterables.


````python
text = "ABC"

for position, character in enumerate(text, start=1):
    print(position, character)

numbers = (10, 20, 30)
for position, value in enumerate(numbers):
    print(position, value)

def generate():
    yield "x"
    yield "y"

for position, value in enumerate(generate()):
    print(position, value)
````

## 6. It is lazy

`enumerate()` does not need to build a list of all `(index, value)` pairs first.

The pairs are produced as the iterator is consumed.


````python
items = ["A", "B", "C"]

pairs = enumerate(items)

print(next(pairs))
print(next(pairs))
print(next(pairs))

# next(pairs) would raise StopIteration
````

## 7. Practical pattern: find an item

`enumerate()` is useful when we need the position of a matching value.


````python
names = ["Maya", "Asha", "Ravi", "Zoe"]

for index, name in enumerate(names):
    if name == "Ravi":
        print("Found at index:", index)
        break
````

## 8. Practical pattern: update items

For a mutable list, `enumerate()` gives the index when we need to replace elements.


````python
numbers = [1, 2, 3, 4]

for i, value in enumerate(numbers):
    numbers[i] = value * 10

print(numbers)
````

## 9. Practical pattern: compare neighboring items

The index can help when we need to look at another position, although `zip()` or other iteration patterns may be clearer depending on the problem.


````python
values = [10, 20, 15, 30]

for i, value in enumerate(values):
    if i > 0 and value > values[i - 1]:
        print(f"{value} increased after {values[i - 1]}")
````

## 10. Equivalent mental model

The documentation describes behavior equivalent to:

```python
def enumerate(iterable, start=0):
    n = start
    for elem in iterable:
        yield n, elem
        n += 1
```

This is a teaching model: the real implementation is a built-in iterator type.


````python
def my_enumerate(iterable, start=0):
    n = start
    for elem in iterable:
        yield n, elem
        n += 1

print(list(my_enumerate(["a", "b"], start=5)))
print(list(enumerate(["a", "b"], start=5)))
````

## 11. `enumerate()` vs `zip()`

Use:
- `enumerate(items)` when we need **index + item**;
- `zip(a, b)` when we need **corresponding values from multiple iterables**.

They can also be combined when necessary.


````python
names = ["Asha", "Ravi", "Zoe"]
scores = [91, 84, 95]

for index, (name, score) in enumerate(zip(names, scores), start=1):
    print(index, name, score)
````
