# Functions, `*args` / `**kwargs`, Lambda & Functional Programming


## 1. Functions: the foundation

A function is a callable block of reusable behavior. `def` introduces a function definition.

**Key vocabulary**
- **Parameter**: a name in the function definition.
- **Argument**: a value supplied when the function is called.
- **Return value**: the object produced by `return`.
- **Default parameter**: a parameter with a value used when the caller omits it.
- Python functions are objects, so they can be stored in variables, passed to other functions, and returned from functions.


````python
def add(a, b):
    """Return the sum of two numbers."""
    return a + b

result = add(10, 20)
print(result)
print(add.__name__)
print(add.__doc__)
````

### Parameter styles

Python supports:
1. positional-or-keyword parameters;
2. positional-only parameters, before `/`;
3. keyword-only parameters, after `*`;
4. variable positional parameters, `*args`;
5. variable keyword parameters, `**kwargs`.

The `/` and `*` markers are part of Python's function signature syntax and let API authors make calling conventions explicit.


````python
def example(pos_only, /, normal, *, keyword_only):
    return pos_only, normal, keyword_only

print(example(1, 2, keyword_only=3))

# These would raise TypeError:
# example(pos_only=1, normal=2, keyword_only=3)
# example(1, 2, 3)
````

## 2. `*args`

`*args` collects extra **positional** arguments into a tuple.

Think:

`caller -> many positional values -> tuple named args`

The name `args` is conventional, not required. `*values` works too.


````python
def total(*args):
    print("args:", args)
    print("type:", type(args).__name__)
    return sum(args)

print(total(10, 20, 30))
print(total())
````

### `*args` with normal and keyword-only parameters

Parameters after `*args` are keyword-only. This is useful when you want to accept an arbitrary number of values but keep configuration explicit.


````python
def report(title, *values, unit="items"):
    return f"{title}: {sum(values)} {unit}"

print(report("Sales", 10, 20, 30))
print(report("Distance", 5, 10, unit="km"))
````

## 3. `**kwargs`

`**kwargs` collects extra **keyword arguments** into a dictionary.

Think:

`caller -> many name=value pairs -> dict named kwargs`

Again, `kwargs` is only a convention; `**options` is equally valid.


````python
def show_profile(**kwargs):
    print("kwargs:", kwargs)
    print("type:", type(kwargs).__name__)

show_profile(name="Amrutha", role="Intern", topic="Python")
````

### Combining `*args` and `**kwargs`

The usual order is:

`normal parameters, *args, keyword-only parameters, **kwargs`

`*args` receives extra positional arguments; `**kwargs` receives extra keyword arguments.


````python
def inspect_call(required, *args, mode="normal", **kwargs):
    print("required =", required)
    print("args =", args)
    print("mode =", mode)
    print("kwargs =", kwargs) 

inspect_call(
    "A", # required
    10, 20, # args
    mode="debug", # keyword argument
    user="Sam", # kwargs
    active=True, # kwargs
)
````

## 4. Packing vs unpacking

The same `*` / `**` syntax appears in function calls, but there it **unpacks** an existing collection.

- `*sequence` -> positional arguments
- `**mapping` -> keyword arguments

This is the reverse of collection inside a function definition.


````python
def power(base, exponent):
    return base ** exponent

values = (2, 5)
options = {"base": 3, "exponent": 4}

print(power(*values))
print(power(**options))
````

### A very common pattern

Build arguments dynamically, then unpack them:

```python
args = [...]
kwargs = {...}
function(*args, **kwargs)
```

This is useful for wrappers, adapters, configuration-driven code, and forwarding arguments.


````python
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

args = ["Amrutha"]
kwargs = {"greeting": "Welcome"}

print(greet(*args, **kwargs))
````

## 5. `*args` vs `**kwargs`

| Feature | `*args` | `**kwargs` |
|---|---|---|
| Collects | extra positional arguments | extra keyword arguments |
| Collection type | tuple | dict |
| Example | `f(1, 2, 3)` | `f(a=1, b=2)` |
| Definition | `def f(*args)` | `def f(**kwargs)` |
| Call unpacking | `f(*values)` | `f(**mapping)` |

**Memory trick:** one `*` → positional; two `**` → keyword.


## 6. Lambda expressions

A lambda creates a small anonymous function.

Syntax:

```python
lambda parameters: expression
```

Important restriction: the body is a **single expression**, not a sequence of statements.

A lambda is not a fundamentally different kind of function; it is a compact way to create a function object.


````python
square = lambda x: x * x

print(square(6))
print(type(square).__name__)
````

### Lambda with multiple parameters

```python
lambda a, b: a + b
```

The expression's value becomes the return value.


````python
combine = lambda first, second: f"{first}-{second}"

print(combine("A", "B"))
````

## 7. Practical lambda use cases

Lambdas are especially common when an API asks for a small function, such as a `key=` function for sorting.

Prefer `def` when the logic needs multiple steps, documentation, testing, or a meaningful reusable name.


````python
people = [
    {"name": "Zoe", "age": 31},
    {"name": "Asha", "age": 28},
    {"name": "Mina", "age": 24},
]

by_age = sorted(people, key=lambda person: person["age"])
print(by_age)

by_name = sorted(people, key=lambda person: person["name"])
print(by_name)
````

## 8. First-class functions and higher-order functions

Python functions are first-class objects:
- assign them to variables;
- put them in collections;
- pass them as arguments;
- return them from other functions.

A **higher-order function** accepts a function, returns a function, or both.


````python
def apply_twice(func, value):
    return func(func(value))

def increment(x):
    return x + 1

print(apply_twice(increment, 10))
````

## 9. Functional-style built-ins

Useful built-ins to know alongside lambdas and higher-order functions:
- `map(function, iterable)` : lazily transforms items;
- `filter(function, iterable)` : lazily keeps selected items;
- `sorted(iterable, key=..., reverse=...)` : returns a new sorted list;
- `sum(iterable)` : totals values;
- `any(iterable)` : true if at least one item is truthy;
- `all(iterable)` : true if every item is truthy;
- `zip(*iterables)` : combines corresponding items.

These are often clearer when combined with named functions or small lambdas.


````python
numbers = [1, 2, 3, 4, 5]

doubled = list(map(lambda x: x * 2, numbers))
evens = list(filter(lambda x: x % 2 == 0, numbers))

print(doubled)
print(evens)
print(sum(numbers))
print(any(x > 4 for x in numbers))
print(all(x > 0 for x in numbers))
print(list(zip(["A", "B", "C"], [10, 20, 30])))
````

## 11. Practice exercises

1. Write `average(*numbers)` that returns the arithmetic mean.
2. Write `build_url(base, **params)` that returns a URL-like string.
3. Write a function with one positional-only parameter and two keyword-only parameters.
4. Sort a list of dictionaries by a nested field using `key=`.
5. Create a function factory that returns a lambda.

````python
#1
def average(*numbers):
    return sum(numbers)/len(numbers)
print(average(3,3,6))
````

````python
# 2
def build_url(base,**params):
    if not params:
        return base
    query_string="&".join(f"{key}={value}" for key,value in params.items())
    return f"{base}?{query_string}"
print(build_url("www.example.com",name="Amrutha",age="21"))
````

````python
#3
def student(name,/,*,age=19,branch="CSE"):
    return f"Name={name},Age={age},Branch={branch}"
print(student("Amrutha"))
print(student("Riya",age=20,branch="ISE"))
````

````python
#4
data=[
    {"name":"Alice","marks":{"maths":85,"science":90}},
    {"name":"Bob","marks":{"maths":90,"science":75}}
]
res=sorted(data,key=lambda x:x["marks"]["science"],reverse=True)
for i in res:
    print(i)
````

````python
#5
def multiplier_factory(n):
    return lambda x:x*n
double=multiplier_factory(2)
triple=multiplier_factory(3)
print(double(5))
print(triple(5))
````
