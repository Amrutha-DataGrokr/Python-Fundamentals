# Python Logging



## 1. Why use logging?

Logging is used to record what an application is doing. Unlike `print()`, logging supports severity levels and configurable output.

| Level | Purpose |
|---|---|
| `DEBUG` | Detailed debugging information |
| `INFO` | Normal application activity |
| `WARNING` | Something unexpected, but the program can continue |
| `ERROR` | An operation failed |
| `CRITICAL` | A serious failure |

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.debug("Debug message")
logging.info("Program started")
logging.warning("Something unexpected happened")
logging.error("An error occurred")
```

## 2. Creating a logger

The usual pattern is `logging.getLogger(__name__)`. It gives each module its own named logger.

```python
import logging

logger = logging.getLogger(__name__)
logger.warning("This is a warning")
logger.error("This is an error")
```

## 3. Logging level

The level controls which messages are processed. With `INFO`, `DEBUG` messages are ignored.

```python
import logging

logger = logging.getLogger("demo")
logger.setLevel(logging.INFO)

logger.debug("Not shown")
logger.info("Shown")
logger.warning("Also shown")
```

## 4. `basicConfig()`

`basicConfig()` is the simplest way to configure logging. Common options include `level`, `format`, `filename`, and `encoding`.

```python
import logging

logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s | %(levelname)s | %(message)s",
    force=True
)

logging.debug("Debug information")
logging.info("Application started")
logging.warning("Warning message")
```

## 5. Logging variables

Pass values as additional arguments. This avoids formatting the message when the level is disabled.

```python
import logging

logging.basicConfig(level=logging.INFO, force=True)
logger = logging.getLogger(__name__)

name = "Amrutha"
count = 5
logger.info("User %s processed %d items", name, count)
```

## 6. Logging to a file

Use `filename` in `basicConfig()` to write logs to a file.

```python
import logging

logging.basicConfig(
    filename="application.log",
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s",
    encoding="utf-8",
    force=True
)

logging.info("Application started")
logging.warning("Low disk space")
print("Log written to application.log")
```

## 7. Logging exceptions

Inside an `except` block, `logger.exception()` records the exception and its traceback.

```python
import logging

logging.basicConfig(level=logging.ERROR, force=True)
logger = logging.getLogger(__name__)

try:
    result = 10 / 0
except ZeroDivisionError:
    logger.exception("Division failed")
```

## 8. Logger, Handler, Formatter

- **Logger** → creates logging events.
- **Handler** → sends logs to a destination such as the console or a file.
- **Formatter** → controls how the log is displayed.

For simple programs, `basicConfig()` is usually enough.

```python
import logging

logger = logging.getLogger("handler_demo")
logger.setLevel(logging.INFO)
logger.handlers.clear()
logger.propagate = False

handler = logging.StreamHandler()
handler.setFormatter(logging.Formatter("%(levelname)s | %(message)s"))
logger.addHandler(handler)

logger.info("Message sent through a handler")

logger.removeHandler(handler)
handler.close()
```
