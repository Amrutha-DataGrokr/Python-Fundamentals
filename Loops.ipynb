# Loops & Iteration in Python

The central idea: iteration

**Iteration** means repeatedly obtaining items from an iterable until there are no more items.

An **iterable** is an object from which Python can obtain items one at a time. Common examples:

- `list`
- `tuple`
- `str`
- `dict`
- `set`
- `range`
- files
- many custom objects

A `for` loop does not require a list or a numeric sequence. It works with an iterable.

The important mental model is:

```text
iterable → iterator → next item → next item → ... → StopIteration
```

A `for` loop hides most of this machinery for you.


````python
# Many different objects can be iterated over
for item in [10, 20, 30]:
    print(item)

for character in "abc":
    print(character)

for number in range(3):
    print(number)

for key in {"a": 1, "b": 2}:
    print(key)
````

### Key distinction: iterable vs iterator

- **Iterable:** can produce an iterator, usually through `iter(obj)`.
- **Iterator:** produces the next value through `next(iterator)` and remembers its current position.

A `for` loop obtains an iterator and repeatedly asks it for the next item until iteration ends.


````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(type(numbers))
print(type(iterator))

print(next(iterator))
print(next(iterator))
print(next(iterator))

# The iterator is now exhausted.
try:
    print(next(iterator))
except StopIteration:
    print("Iteration is finished")
````

## 1. `for` loops

Python's `for` statement iterates over the items of an iterable in order.

General form:

```python
for variable in iterable:
    # body
```

The loop variable receives one item at a time.

**Important:** Python's `for` loop is not fundamentally a "counting loop". Counting is one possible use case, usually through `range()`. The general operation is **iterate over items**.

Official documentation:  
https://docs.python.org/3/tutorial/controlflow.html#for-statements


````python
# Direct iteration over values
fruits = ["apple", "banana", "mango"]

for fruit in fruits:
    print(fruit)
````

````python
# Iterating over a string
for character in "Python":
    print(character)
````

````python
# Iterating over a dictionary
person = {"name": "Amrutha", "age": 21}

for key in person:
    print(key)

print("--- values ---")
for value in person.values():
    print(value)

print("--- key/value pairs ---")
for key, value in person.items():
    print(key, value)
````

### `for` loop and unpacking

If each item is itself an iterable with the expected number of elements, the loop target can unpack it.

This is why this pattern works:

```python
for key, value in dictionary.items():
    ...
```

Each item from `.items()` is a two-element tuple.


````python
pairs = [("a", 1), ("b", 2), ("c", 3)]

for letter, number in pairs:
    print(letter, number)
````

### Looping over a collection while modifying it

Python's tutorial warns that changing a collection while iterating over that same collection can be tricky.

Prefer either:
1. iterating over a copy, or
2. building a new collection.

This matters especially for dictionaries and lists.


````python
# Safer: iterate over a copy
users = {
    "A": "active",
    "B": "inactive",
    "C": "active",
}

for user, status in users.copy().items(): # iterate over a copy of the dictionary
    if status == "inactive":
        del users[user]

print(users)
````

## 2. `while` loops

A `while` loop repeatedly executes its body **while a condition is true**.

General form:

```python
while condition:
    # body
```

Unlike `for`, the stopping condition is explicitly controlled by the program.

Official documentation:  
https://docs.python.org/3/tutorial/controlflow.html#while-statements


````python
count = 0

while count < 5:
    print(count)
    count += 1
````

### The three parts of a typical `while` loop

1. **Initialization** : establish the state used by the condition.
2. **Condition** : decide whether another iteration should happen.
3. **Progress/update** : change state so that the loop can eventually terminate.

If the update is missing or incorrect, an infinite loop can occur.


````python
# Example: process a value until it reaches a boundary
n = 100

while n > 1:
    print(n)
    n //= 2

print("finished:", n)
````

### `for` vs `while`

| Question | `for` | `while` |
|---|---|---|
| Main idea | Iterate over an iterable | Repeat while condition is true |
| Best for | Known collection/sequence of items | Condition-driven repetition |
| Counter required? | No | Often |
| Natural termination | Iterable becomes exhausted | Condition becomes false |
| Infinite loop risk | Lower | Higher if state is not updated |

A `for` loop is usually preferable when you already have an iterable to traverse.


````python
# Same conceptual task, two styles

# for: iterate over the values
for number in [1, 2, 3, 4, 5]:
    print(number)

print("---")

# while: explicitly control the index
numbers = [1, 2, 3, 4, 5]
i = 0

while i < len(numbers):
    print(numbers[i])
    i += 1
````

## 3. `range()`

`range()` represents an immutable sequence of numbers, commonly used for iteration.

Forms:

```python
range(stop)
range(start, stop)
range(start, stop, step)
```

Rules:

- `start` defaults to `0`.
- `step` defaults to `1`.
- `stop` is **exclusive**.
- `step` can be negative.
- `step` cannot be zero (`ValueError`).

Official documentation:  
https://docs.python.org/3/library/stdtypes.html#ranges  
and  
https://docs.python.org/3/tutorial/controlflow.html#the-range-function


````python
print(list(range(5)))
print(list(range(2, 7)))
print(list(range(2, 10, 2)))
print(list(range(10, 2, -2)))
````

````python
# stop is exclusive
print(list(range(5)))       # 0, 1, 2, 3, 4
print(list(range(1, 5)))    # 1, 2, 3, 4
````

````python
# Negative step
for n in range(5, 0, -1):
    print(n)
````

````python
# step cannot be zero
try:
    range(1, 5, 0)
except ValueError as error:
    print(type(error).__name__, error)
````

### `range()` does not build a list of all numbers

This is an important space-complexity point.

```python
r = range(1_000_000_000)
```

does **not** create a billion-element list. A `range` object represents the arithmetic progression and produces values when iterated.

So:

- `range(n)` itself uses **O(1) auxiliary space**.
- Iterating through `range(n)` takes **O(n) time**.
- `list(range(n))` materializes all values and therefore uses **O(n) space**.


````python
import sys

r = range(1_000_000)
lst = list(range(1_000_000))

print("range object size:", sys.getsizeof(r)) # size of the range object in bytes
print("list object size:", sys.getsizeof(lst)) # size of the list object in bytes
````

### `range()` supports sequence operations

A range is a sequence, so you can use indexing, slicing, membership tests, and `len()`.

Do not confuse **logical size** with **memory used**.


````python
r = range(10, 20, 2)

print("range:", r)
print("length:", len(r))
print("first:", r[0])
print("last:", r[-1])
print("slice:", r[1:3])
print("membership:", 14 in r)
````

## 4. `len()`

`len(obj)` returns the number of items in an object.

Official documentation:  
https://docs.python.org/3/library/functions.html#len

Common objects supporting `len()` include:

- strings
- lists
- tuples
- dictionaries
- sets
- ranges
- many user-defined containers


````python
print(len("Python"))
print(len([10, 20, 30]))
print(len((10, 20)))
print(len({"a": 1, "b": 2}))
print(len({1, 2, 3}))
print(len(range(100)))
````

### `len()` and complexity

For Python's standard built-in containers, `len()` is generally designed to be an O(1) operation because the container maintains its size.

Do not assume every custom object has O(1) `len()`. A custom type can implement `__len__()` with arbitrary work.

Also note:

```python
len(dictionary)
```

counts **keys**, and:

```python
len(set)
```

counts **elements**.


````python
data = {"a": 10, "b": 20, "c": 30}

print(len(data))
print(len(data.keys()))
print(len(data.values()))
print(len(data.items()))
````

## 5. `range()` + `len()`: indexing patterns

A common pattern is:

```python
for i in range(len(items)):
    print(i, items[i])
```

This is useful when you genuinely need the **index**.

However, if you need both index and value, `enumerate()` is normally clearer.


````python
words = ["Mary", "had", "a", "little", "lamb"]

for i in range(len(words)):
    print(i, words[i])
````

### Choosing between the two

| Need | Pattern |
|---|---|
| Values only | `for value in items` |
| Index + value | `for i, value in enumerate(items)` |
| Only indices | `for i in range(len(items))` |
| Numeric progression unrelated to a collection | `for i in range(...)` |

The Python tutorial specifically points out that `enumerate()` is often more convenient than `range(len(...))`.


## 6. `zip()`

`zip(*iterables, strict=False)` iterates over multiple iterables in parallel and produces tuples containing corresponding items.

Example:

```python
zip([1, 2, 3], ["a", "b", "c"])
```

produces:

```text
(1, "a")
(2, "b")
(3, "c")
```

Official documentation:  
https://docs.python.org/3/library/functions.html#zip


````python
names = ["Alice", "Bob", "Charlie"]
scores = [85, 92, 78]

for name, score in zip(names, scores):
    print(name, score)
````

### `zip()` is lazy

`zip()` returns an iterator. It does not immediately build all output tuples.

Therefore this is useful for large inputs:

```python
for a, b in zip(first, second):
    ...
```

rather than unnecessarily materializing:

```python
list(zip(first, second))
```


````python
a = [1, 2, 3]
b = ["x", "y", "z"]

z = zip(a, b)

print(z)
print(type(z))
print(next(z)) # returns the first pair (1, 'x') 
print(next(z)) # returns the second pair (2, 'y')
print(list(z))# returns the remaining pairs [(3, 'z')]
````

### Different lengths: the default behavior

By default, `zip()` stops when the **shortest iterable** is exhausted.

This can be useful, but it can also hide a data-length bug.


````python
numbers = [1, 2, 3]
letters = ["a", "b", "c", "d", "e"]

print(list(zip(numbers, letters)))
````

### `strict=True`

When the iterables are expected to have the same length, use:

```python
zip(a, b, strict=True)
```

If lengths differ, Python raises `ValueError`.

This option was added in Python 3.10.


````python
a = [1, 2, 3]
b = ["a", "b", "c"]

print(list(zip(a, b, strict=True)))
````

````python
a = [1, 2, 3]
b = ["a", "b", "c", "d"]

try:
    print(list(zip(a, b, strict=True)))
except ValueError as error:
    print(type(error).__name__, error)
````

### `zip()` with zero or one iterable

- `zip()` with no arguments → empty iterator.
- `zip(one_iterable)` → tuples containing one element each.


````python
print(list(zip()))
print(list(zip([1, 2, 3])))
````

### Unzipping with `*`

If you have paired data, `zip(*pairs)` can reverse the operation.

This is a useful application of iterable unpacking.


````python
pairs = [("Alice", 90), ("Bob", 85), ("Charlie", 92)]

names, scores = zip(*pairs)

print(names)
print(scores)
````

### Complexity of `zip()`

Suppose there are `k` iterables and the shortest one contains `n` items.

Traversing the resulting zip:

- Time: **O(k × n)** in the simple model, because each produced tuple obtains one item from each iterable.
- Auxiliary space of the `zip` iterator: **O(k)** for references to the input iterators.
- `list(zip(...))`: output storage is **O(k × n)**.

For the common case of two iterables, traversal is O(n).


## 7. `iter()` and `next()`: the machinery behind iteration

These are especially useful for understanding what `for` is doing.

### `iter(obj)`

Returns an iterator for an iterable.

### `next(iterator)`

Returns the next item from an iterator.

When the iterator is exhausted, `next()` raises `StopIteration`.

Official documentation:
- `iter()`: https://docs.python.org/3/library/functions.html#iter
- `next()`: https://docs.python.org/3/library/functions.html#next


````python
items = ["a", "b", "c"]

it = iter(items)

print(next(it))
print(next(it))
print(next(it))

try:
    print(next(it))
except StopIteration:
    print("No more items")
````

## 8. `break`

`break` immediately terminates the **innermost** enclosing `for` or `while` loop.

Official documentation:  
https://docs.python.org/3/tutorial/controlflow.html#break-and-continue-statements


````python
for number in range(10):
    if number == 5:
        break
    print(number)

print("loop ended")
````

## 9. `continue`

`continue` skips the rest of the current iteration and proceeds to the next iteration.


````python
for number in range(10):
    if number % 2 == 0:
        continue
    print(number)
````

### `break` vs `continue`

| Statement | Effect |
|---|---|
| `break` | Exit the loop completely |
| `continue` | Skip the current iteration and continue looping |

Both affect only the innermost enclosing loop.


## 10. Loop `else`

Python's `for` and `while` loops can have an `else` clause.

The `else` block runs when the loop finishes **normally**, meaning it was not terminated by `break`.

This is especially useful for search patterns.


````python
# Search: else means "not found"
target = 7

for number in range(5):
    if number == target:
        print("Found")
        break
else:
    print("Not found")
````

````python
target = 3

for number in range(5):
    if number == target:
        print("Found")
        break
else:
    print("Not found")
````

## 11. Combining `enumerate()` and `zip()`

This is a common real-world pattern when you need:

- an index,
- an item from one iterable,
- and a corresponding item from another iterable.


````python
names = ["Alice", "Bob", "Charlie"]
scores = [90, 85, 92]

for index, (name, score) in enumerate(zip(names, scores), start=1):
    print(index, name, score)
````

Notice the unpacking structure:

```python
for index, (name, score) in enumerate(zip(names, scores)):
```

The `zip()` item is `(name, score)`, and `enumerate()` adds the index:

```text
(index, (name, score))
```


## 12. `zip()` vs indexing

You may sometimes see:

```python
for i in range(len(names)):
    print(names[i], scores[i])
```

This works when the sequences are indexable.

But:

```python
for name, score in zip(names, scores):
    print(name, score)
```

expresses the intention directly: **iterate over corresponding values**.

`zip()` is also more general because it works with iterables that are not necessarily indexable.


````python
names = ["Alice", "Bob", "Charlie"]
scores = [90, 85, 92]

# Index-based
for i in range(len(names)):
    print(names[i], scores[i])

print("---")

# Iterable-based
for name, score in zip(names, scores):
    print(name, score)
````

## 13. Nested loops

A loop inside another loop is a **nested loop**.

If the outer loop runs `n` times and the inner loop runs `m` times for every outer iteration, the total number of body executions is approximately:

**O(n × m)**

If both are based on the same `n`:

**O(n²)**



````python
for i in range(3):
    for j in range(4):
        print(i, j)
````

### Complexity example

```python
for i in range(n):
    for j in range(n):
        work()
```

The inner body runs:

```text
n × n = n²
```

times.

Therefore, if `work()` is O(1), total time is **O(n²)**.

The two loop variables themselves require O(1) auxiliary space, assuming `work()` does not allocate growing storage.


## 14. Sequential loops vs nested loops

Two loops one after another are **not automatically O(n²)**.

```python
for x in items:
    work(x)

for y in items:
    work(y)
```

Each loop is O(n), so:

```text
O(n) + O(n) = O(n)
```

Big-O ignores constant factors.


````python
# Two sequential linear passes
numbers = list(range(100))

for x in numbers:
    pass

for y in numbers:
    pass
````

## 15. Complexity: a practical cheat sheet

Assume `n` is the number of items.

| Pattern | Time | Auxiliary space* |
|---|---:|---:|
| `for x in items` | O(n) | O(1) |
| `for x in range(n)` | O(n) | O(1) |
| `for i, x in enumerate(items)` | O(n) | O(1) |
| `for a, b in zip(A, B)` | O(min(n, m)) | O(1) iterator overhead |
| `list(range(n))` | O(n) | O(n) |
| `list(enumerate(items))` | O(n) | O(n) |
| `list(zip(A, B))` | O(min(n, m)) | O(min(n, m)) |
| nested `n × n` loops | O(n²) | O(1) if no growing storage |
| two sequential `n` loops | O(n) | O(1) if no growing storage |

\*Auxiliary space excludes the input itself and excludes output storage unless stated.

**Important:** exact complexity can depend on the underlying iterable and the operation performed inside the loop.


## 16. A subtle complexity issue: `in`

Membership tests can have different costs depending on the container.

For example, a membership test in a list generally scans items:

```python
x in my_list
```

→ O(n) worst case.

A set/dictionary membership test is generally O(1) average case.

This means that putting a linear membership test inside a loop can accidentally produce O(n²) code.


````python
# Potential O(n^2) pattern
items = list(range(100))

for x in items:
    if x in items:
        pass
````

A useful rule:

> When analyzing a loop, do not count only the loop itself. Analyze the work performed **inside** each iteration.


# Final mental model

Keep these relationships in mind:

```text
ITERABLE
   │
   ├── for item in iterable
   │       │
   │       └── Python obtains an ITERATOR
   │                 │
   │                 ├── next(...)
   │                 ├── next(...)
   │                 └── StopIteration
   │
   ├── range(...)      → numeric iterable/sequence
   ├── enumerate(...)  → iterator of (index, value)
   └── zip(...)        → iterator of tuples from parallel iterables
```

And the most useful selection rule:

```text
Need values?             → for value in items
Need index + value?      → enumerate(items)
Need parallel values?    → zip(a, b)
Need numeric progression?→ range(...)
Need condition control?  → while
Need early exit?         → break
Need skip iteration?     → continue
Need "not found"?        → loop else
```
