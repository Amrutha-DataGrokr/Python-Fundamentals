## 1. Creating Dictionaries

A dictionary is a mutable mapping of keys to values. Keys must be hashable.

**Official:** https://docs.python.org/3/library/stdtypes.html#mapping-types-dict

```python
empty = {}
also_empty = dict()
student = {"name": "Amrutha", "age": 21, "branch": "CSE"}
from_pairs = dict([("a", 1), ("b", 2)])
from_kwargs = dict(a=1, b=2)

print(empty)
print(also_empty)
print(student)
print(from_pairs)
print(from_kwargs)
```

## 2. Key Lookup, Assignment, and Deletion

`d[key]` retrieves a value and raises `KeyError` when the key is absent. Assignment can add or replace a key-value pair. `del` removes a key.

```python
d = {"a": 10, "b": 20}

print(d["a"])

d["a"] = 100
d["c"] = 30
print(d)

del d["b"]
print(d)

try:
    print(d["missing"])
except KeyError as e:
    print(type(e).__name__, e)
```

## 3. `get()`

Returns the value for a key. If the key is absent, it returns `None` or the supplied default instead of raising `KeyError`.

```python
d = {"a": 10}

print(d.get("a"))
print(d.get("missing"))
print(d.get("missing", 0))
```

## 4. `keys()`, `values()`, and `items()`

These return dynamic view objects over the dictionary's keys, values, and key-value pairs.

```python
d = {"a": 1, "b": 2}

keys = d.keys()
values = d.values()
items = d.items()

print(keys)
print(values)
print(items)

d["c"] = 3
print(keys)
print(values)
print(items)
```

## 5. `pop()`

Removes the specified key and returns its value. A default can be supplied for a missing key.

```python
d = {"a": 10, "b": 20}

print(d.pop("a"))
print(d)
print(d.pop("missing", 0))

try:
    d.pop("missing")
except KeyError as e:
    print(type(e).__name__, e)
```

## 6. `popitem()`

Removes and returns a key-value pair as a two-item tuple. It follows last-in, first-out order.

```python
d = {"a": 1, "b": 2, "c": 3}

print(d.popitem())
print(d)
```

## 7. `clear()`

Removes all dictionary items and returns `None`.

```python
d = {"a": 1, "b": 2}
result = d.clear()

print(d)
print(result)
```

## 8. `copy()`

Returns a shallow copy of the dictionary.

```python
d = {"a": [1, 2]}
c = d.copy()

print(d is c)
print(d == c)

c["a"].append(3)
print(d)
print(c)
```

## 9. `update()`

Updates the dictionary with key-value pairs from another mapping or iterable of pairs, or from keyword arguments.

```python
d = {"a": 1, "b": 2}

result = d.update({"b": 20, "c": 30})
print(d)
print(result)

d.update(d=40)
print(d)
```

## 10. `setdefault()`

Returns the value for a key. If the key is absent, it inserts the key with the supplied default and returns that default.

```python
d = {"a": 1}

print(d.setdefault("a", 100))
print(d)

print(d.setdefault("b", 200))
print(d)
```

## 11. `fromkeys()`

Creates a new dictionary with keys from an iterable and the same value for each key.

```python
keys = ["a", "b", "c"]

d = dict.fromkeys(keys, 0)
print(d)

empty_values = dict.fromkeys(keys)
print(empty_values)
```

## 12. Dictionary Iteration

Iterating directly over a dictionary yields keys. Use `values()` for values and `items()` for key-value pairs.

```python
d = {"a": 1, "b": 2}

for key in d:
    print("key:", key)

for value in d.values():
    print("value:", value)

for key, value in d.items():
    print(key, value)
```

## 13. Membership

`in` and `not in` test dictionary keys, not values.

```python
d = {"a": 1, "b": 2}

print("a" in d)
print("x" not in d)
print(1 in d)
print(1 in d.values())
```

## 14. Dictionary Ordering

Dictionaries preserve insertion order. Updating an existing key does not move it; removing and re-inserting a key adds it at the end.

```python
d = {"a": 1, "b": 2, "c": 3}
d["b"] = 20
print(list(d))

del d["b"]
d["b"] = 200
print(list(d))
```

## 15. Dictionary Comprehensions

A dictionary comprehension creates a new dictionary from an iterable.

```python
squares = {x: x * x for x in range(5)}
even_squares = {x: x * x for x in range(10) if x % 2 == 0}

print(squares)
print(even_squares)
```

## 16. Hashable Keys

Dictionary keys must be hashable. Values do not have this restriction.

```python
d = {(1, 2): "tuple key", "numbers": [1, 2, 3]}
print(d)

try:
    invalid = {[1, 2]: "value"}
except TypeError as e:
    print(type(e).__name__, e)
```

## 17. Complexity Quick Reference

| Operation | Typical complexity |
|---|---:|
| Lookup `d[key]` | O(1) average |
| Assignment | O(1) average |
| Membership | O(1) average |
| `get()` | O(1) average |
| `pop()` | O(1) average |
| `popitem()` | O(1) average |
| `len()` | O(1) |
| Iteration | O(n) |
| `copy()` | O(n) |
| `update()` | O(m) average for m inserted/updated entries |

Dictionary average-case complexity relies on hash-table behavior; avoid treating it as a universal worst-case guarantee.

## 18. Memory / Storage Notes

- Dictionaries store mappings between keys and values.
- Keys must be hashable.
- Values can be mutable.
- `copy()` is shallow.
- Dictionary views (`keys()`, `values()`, `items()`) provide dynamic views rather than independent copied collections.
- Dictionaries preserve insertion order.

## 19. Quick Revision

**Dictionary = mutable + key-value mapping + unique keys + hashable keys**

Use a dictionary when data should be accessed by a meaningful key rather than by position.
