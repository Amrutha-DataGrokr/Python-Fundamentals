# `def`, Parameters, Arguments, Defaults & Annotations



## 1. What is a function?

A function is a reusable unit of behavior. In Python, a function is also an **object**.

A function can:

- receive input through parameters,
- execute statements,
- return a result,
- be assigned to another name,
- be passed to another function,
- be returned by another function.

The core mental model is:

> The `def` statement creates/binds a function object. Calling the function executes its body.

```python
def add(a, b):
    return a + b
```

The expression `a + b` is not executed when Python merely defines the function. It executes when `add(...)` is called.

The official Python Tutorial states that `def` introduces a function definition, followed by the function name and formal parameters; the body is indented.

````python
def add(a, b):
    return a + b

print(add)
print(type(add))
print(add(2, 3))
````

## 2. Function definition vs function call

These are different:

```python
def greet(name):
    return f"Hello {name}"
```

This **defines** the function.

```python
greet("Amrutha")
```

This **calls** the function.

Also:

```python
greet
```

refers to the function object, whereas:

```python
greet()
```

calls it.

This distinction becomes important for callbacks, decorators, `sorted(key=...)`, `map`, and higher-order functions.

````python
def greet(name):
    return f"Hello {name}"

print("Function object:", greet)
print("Call result:", greet("Amrutha"))
````

## 3. Anatomy of a `def` statement

A basic function looks like:

```text
def function_name(parameters):
    body
```

An annotated function can look like:

```text
def function_name(parameters) -> return_annotation:
    body
```

For example:

```python
def greet(name: str) -> str:
    return f"Hello, {name}"
```

| Part | Meaning |
|---|---|
| `def` | Keyword introducing a function definition |
| `greet` | Function name |
| `name` | Parameter |
| `: str` | Parameter annotation |
| `-> str` | Return annotation |
| `:` | Ends the function header |
| indented block | Function body |

````python
def greet(name: str) -> str:
    return f"Hello, {name}"

print(greet("Amrutha"))
````

## 4. Parameters vs arguments

This distinction is fundamental.

### Parameter

A parameter is the name written in the function definition:

```python
def greet(name):
    ...
```

`name` is a parameter.

### Argument

An argument is the value supplied when the function is called:

```python
greet("Amrutha")
```

`"Amrutha"` is an argument.

So:

```text
def greet(name):
           ^^^^
        parameter

greet("Amrutha")
      ^^^^^^^^^
        argument
```

**Rule:** parameters belong to the definition; arguments belong to the call.

````python
def introduce(name, age):
    print("name:", name)
    print("age:", age)

introduce("Amrutha", 21)
````

## 5. Required parameters

A parameter without a default is normally required.

```python
def add(a, b):
    return a + b
```

Both `a` and `b` must receive arguments.

If a required argument is missing, Python raises `TypeError`.

````python
def add(a, b):
    return a + b

print(add(2, 3))

try:
    add(2)
except TypeError as e:
    print(type(e).__name__, ":", e)
````

## 6. Default parameters

A parameter can have a default value:

```python
def greet(name, company="ABC"):
    ...
```

Then:

```python
greet("Amrutha")
```

uses `"ABC"`.

But:

```python
greet("Amrutha", "XYZ")
```

overrides the default.

The Language Reference calls these **default parameter values**. If the corresponding argument is omitted, the default is substituted.

````python
def greet(name, company="ABC"):
    return f"{name} works at {company}"

print(greet("Amrutha"))
print(greet("Amrutha", "XYZ"))
print(greet(name="Amrutha", company="XYZ"))
````

## 7. Why can't a required parameter follow a default parameter?

This is invalid:

```python
def f(a=10, b):
    ...
```

A positional parameter with a default cannot be followed by another positional-or-keyword parameter without a default.

Valid:

```python
def f(a, b=10):
    ...
```

Also valid:

```python
def f(a=10, *, b):
    ...
```

because `b` is keyword-only.

````python
def valid(a, b=10):
    return a + b

print(valid(5))

def also_valid(a=10, *, b):
    return a + b

print(also_valid(b=5))
````

## 8. Positional arguments

Arguments can be matched by position:

```python
def subtract(a, b):
    return a - b

subtract(10, 3)
```

Binding:

```text
a is 10
b is 3
```

The order therefore matters.

````python
def subtract(a, b):
    return a - b

print(subtract(10, 3))
print(subtract(3, 10))
````

## 9. Keyword arguments

Arguments can also be supplied by parameter name:

```python
subtract(a=10, b=3)
```

Keyword arguments do not have to appear in the same order as the definition:

```python
subtract(b=3, a=10)
```

This is one reason meaningful parameter names are useful.

````python
def describe(name, age, city):
    return f"{name}, {age}, from {city}"

print(describe(name="Amrutha", age=21, city="Bengaluru"))
print(describe(city="Bengaluru", name="Amrutha", age=21))
````

## 10. Positional + keyword arguments

You can mix them, but positional arguments must come before keyword arguments.

Valid:

```python
describe("Amrutha", age=21, city="Bengaluru")
```

Invalid:

```python
describe(name="Amrutha", 21, city="Bengaluru")
```

Also, a parameter cannot receive a value twice.

````python
def describe(name, age, city):
    return f"{name}, {age}, from {city}"

print(describe("Amrutha", age=21, city="Bengaluru"))

try:
    describe("Amrutha", name="Other", age=21, city="Bengaluru")
except TypeError as e:
    print(type(e).__name__, ":", e)
````

## 11. `return`

`return` sends a value back to the caller and immediately ends the current function call.

```python
def add(a, b):
    return a + b
```

The result can be assigned, printed, or used in another expression.

````python
def add(a, b):
    return a + b

result = add(10, 20)

print(result)
print(result * 2)
print(add(2, 3) + add(4, 5))
````

## 12. What happens without `return`?

A function that reaches the end without returning a value returns `None`.

These also return `None`:

```python
return
```

and:

```python
return None
```

````python
def first():
    print("inside first")

def second():
    return

def third():
    return None

print("first:", first())
print("second:", second())
print("third:", third())
````

## 13. Returning multiple values

Python functions return one object.

When you write:

```python
return a, b
```

the object returned is a tuple:

```python
(a, b)
```

Tuple unpacking can then be used:

```python
x, y = function()
```

````python
def min_max(values):
    return min(values), max(values)

result = min_max([4, 1, 9, 2])

print(result)
print(type(result))

minimum, maximum = min_max([4, 1, 9, 2])
print(minimum, maximum)
````

## 14. Scope and local variables

A function call creates a local namespace.

Assignments inside the function normally create local variables:

```python
x = 100

def test():
    x = 10
```

The `x` inside `test` is different from the global `x`.

Python's name lookup also involves enclosing function scopes, the global namespace, and built-ins.

````python
x = 100

def test():
    x = 10
    print("inside:", x)

test()
print("outside:", x)
````

## 15. Argument binding: the Python mental model

A useful model is that a parameter becomes a local name referring to the object supplied by the caller.

This is often described as **call by object reference** or **call by sharing**.

Two consequences:

1. Rebinding the parameter does not rebind the caller's variable.
2. Mutating a mutable object through the parameter can be visible to the caller.

````python
def rebind(value):
    value = 999

number = 10
rebind(number)
print("after rebind:", number)

def mutate(items):
    items.append(999)

numbers = [1, 2, 3]
mutate(numbers)
print("after mutate:", numbers)
````

## 16. Default expressions are evaluated when the `def` executes

This is an important rule.

```python
x = 5

def f(value=x):
    return value

x = 6
```

`f()` still uses `5`.

The default expression was evaluated when the function definition executed, not every time `f()` is called.

````python
x = 5

def f(value=x):
    return value

x = 6

print(f())
````

## 16. Mutable default argument pitfall

This is a classic Python issue:

```python
def bad(item, items=[]):
    items.append(item)
    return items
```

The default list is created once and can be shared across calls.

A common safe pattern is:

```python
def good(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items
```

The official tutorial explicitly warns about this behavior.

````python
def bad(item, items=[]):
    items.append(item)
    return items

print("Bad:")
print(bad("A"))
print(bad("B"))
print(bad("C"))

def good(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items

print("Good:")
print(good("A"))
print(good("B"))
print(good("C"))
````

## 17. Default values and annotations are separate

Consider:

```python
def greet(name: str = "Guest") -> str:
    ...
```

There are three different concepts:

- `name: str` -> annotation
- `= "Guest"` -> default value
- `-> str` -> return annotation

The annotation does not create the default.
The default does not enforce the annotation.

````python
def greet(name: str = "Guest") -> str:
    return f"Hello {name}"

print(greet())
print(greet("Amrutha"))
print(greet(123))
````

## 18. Function annotations

Function annotations are optional metadata attached to parameters and/or the return value.

Example:

```python
def square(value: int) -> int:
    return value * value
```

Python's runtime does **not** automatically enforce these annotations.

They are useful to:

- static type checkers,
- IDEs,
- linters,
- documentation tools,
- frameworks and libraries,
- developers reading an API.

Annotations are stored in `function.__annotations__`.

````python
def square(value: int) -> int:
    return value * value

print(square(5))
print(square.__annotations__)
````

## 19. `name: str` does not convert the argument

This:

```python
def show_name(name: str):
    ...
```

does not convert `123` to `"123"`.

The annotation is metadata.

If conversion is desired, it must be explicit:

```python
str(123)
```

or implemented by the application's own validation/conversion logic.

````python
def show_name(name: str) -> str:
    return name

value = show_name(123)

print(value)
print(type(value))
print(show_name.__annotations__)
````

## 20. `-> str` does not enforce the return type

This is syntactically valid:

```python
def f() -> str:
    return 123
```

Python does not automatically raise an exception just because the returned object does not match the annotation.

A static type checker may report the mismatch.

Therefore:

> `-> str` is a return annotation, not a cast and not automatic runtime validation.

````python
def f() -> str:
    return 123

result = f()

print(result)
print(type(result))
print(f.__annotations__)
````

````python
def name(value: str) -> str:
    return value

print(name.__annotations__)
print(name.__annotations__["value"])
print(name.__annotations__["return"])
````

## 21. Function metadata and introspection

Useful attributes include:

| Attribute | Meaning |
|---|---|
| `__name__` | Function name |
| `__doc__` | Docstring |
| `__annotations__` | Parameter and return annotations |
| `__defaults__` | Defaults for positional parameters |
| `__kwdefaults__` | Defaults for keyword-only parameters |

The `inspect` module provides `inspect.signature()` to inspect a callable's signature.

````python
import inspect

def example(name: str, age: int = 20, *, active: bool = True) -> str:
    """Describe a user."""
    return f"{name} ({age}) active={active}"

print("__name__:", example.__name__)
print("__doc__:", example.__doc__)
print("__annotations__:", example.__annotations__)
print("__defaults__:", example.__defaults__)
print("__kwdefaults__:", example.__kwdefaults__)
print("signature:", inspect.signature(example))
````

## 22. Functions are first-class objects

Because functions are objects, they can be:

- assigned to variables,
- stored in lists/dictionaries,
- passed as arguments,
- returned from functions.

This is the foundation of callbacks, decorators, and functional programming in Python.

````python
def add(a, b):
    return a + b

operation = add

print(operation(2, 3))
print(operation is add)

operations = [add, lambda x, y: x * y]

print(operations[0](2, 3))
print(operations[1](2, 3))
````

## 23. Functions passed as arguments

A function can receive another function as an argument.

This is possible because functions are first-class objects.

````python
def apply_twice(function, value):
    return function(function(value))

def double(x):
    return x * 2

print(apply_twice(double, 5))
````

## 24. Functions returning functions

A function can create and return another function.

This is important for closures and decorators.

````python
def make_multiplier(factor):
    def multiply(value):
        return value * factor
    return multiply

double = make_multiplier(2)
triple = make_multiplier(3)

print(double(10))
print(triple(10))
````

# Quick-reference

| Concept | Example | Meaning |
|---|---|---|
| Function definition | `def f():` | Defines a function |

| Positional-only | `def f(x, /):` | `x` must be positional |
| Keyword-only | `def f(*, x):` | `x` must be keyword |
| `*args` | `def f(*args):` | Extra positional arguments |
| `**kwargs` | `def f(**kwargs):` | Extra keyword arguments |
| Parameter annotation | `x: int` | Annotation metadata |
| Return annotation | `-> str` | Return annotation metadata |
| Docstring | first string in body | Function documentation |
| `return value` | `return x` | Returns an object |
| no return | `def f(): ...` | Returns `None` |
