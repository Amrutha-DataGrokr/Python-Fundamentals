## 1. Creating Tuples

A tuple is an immutable sequence. Parentheses are optional in many tuple expressions; the comma is what makes a tuple.

**Official:** https://docs.python.org/3/library/stdtypes.html#tuples

```python
empty = ()
one = (10,)
many = (10, 20, 30)
without_parentheses = 10, 20, 30
from_list = tuple([1, 2, 3])
from_range = tuple(range(4))
dict={1: 'a', 2: 'b', 3: 'c'}
from_dict = tuple(dict) # Creates a tuple from the keys of the dictionary dict.
from_dict_values = tuple(dict.values()) # Creates a tuple from the values of the dictionary dict.
from_dict_items = tuple(dict.items()) # Creates a tuple from the key-value pairs of the dictionary dict.

print(empty) # An empty tuple.
print(one, type(one)) # A tuple with a single element. Note: the comma after the element.
print(many) # A tuple with multiple elements.
print(without_parentheses) # A tuple created without parentheses.
print(from_list) # A tuple created from a list.
print(from_range) # A tuple created from a range.
print(from_dict) # A tuple created from the keys of the dictionary.
print(from_dict_values) # A tuple created from the values of the dictionary.
print(from_dict_items) # A tuple created from the key-value pairs of the dictionary.
```

## 2. Indexing and Slicing

Tuples support sequence indexing and slicing. They are zero-based; negative indexes count from the end. Slicing creates a new tuple.

```python
t = (10, 20, 30, 40, 50)

print(t[0])
print(t[-1])
print(t[1:4])
print(t[:3])
print(t[::2])
print(t[::-1])
```

## 3. Sequence Operations

Tuples support membership, concatenation, repetition, indexing, slicing, `len()`, and comparisons.

```python
a = (1, 2, 3)
b = (4, 5)

print(2 in a) # True, because 2 is an element of the tuple a.
print(9 not in a) # True, because 9 is not an element of the tuple a.
print(a + b) # Concatenates the tuples a and b, resulting in a new tuple (1, 2, 3, 4, 5).
print(a * 2) # Repeats the tuple a two times, resulting in a new tuple (1, 2, 3, 1, 2, 3).
print(a[1]) # Accesses the element at index 1 of the tuple a, which is 2.
print(a[1:]) # Slices the tuple a from index 1 to the end, resulting in a new tuple (2, 3).
print(len(a)) # Returns the number of elements in the tuple a, which is 3.
print((1, 2) == (1, 2)) # True, because both tuples have the same elements in the same order.
print((1, 2) < (1, 3)) # True, because the first differing element (2) is less than the corresponding element (3) in the second tuple.
```

## 4. `count()`

Returns the number of occurrences of a value.

```python
t = (1, 2, 2, 3, 2)
print(t.count(2))
print(t.count(9))
```

## 5. `index()`

Returns the first index of a matching value. `start` and `stop` can limit the search. Raises `ValueError` when no match is found.

```python
t = ("a", "b", "c", "b")

print(t.index("b")) # Searches for the index of the first occurrence of "b" in the tuple t. 
print(t.index("b", 2)) # Searches for the index of the first occurrence of "b" in the tuple t, starting from index 2.
# print(t.index("x"))  # ValueError
```

## 6. Tuple Packing and Unpacking

Packing collects values into a tuple. Unpacking assigns tuple elements to variables. Extended unpacking uses `*`.

```python
point = 10, 20
x, y = point

values = (1, 2, 3, 4) 
# Unpacking with starred expressions allows you to capture multiple elements in a single variable. 
# The starred variable will be a list containing all the elements that are not assigned to other variables.
first, *middle, last = values 

print(point)
print(x, y)
print(first, middle, last)
```

## 7. Immutability

Tuple item assignment is not allowed. However, a tuple can contain mutable objects, and those contained objects can still be modified.

```python
t = (1, 2, 3)

try:
    t[0] = 100
except TypeError as e:
    print(type(e).__name__, e) # TypeError: 'tuple' object does not support item assignment

nested = ([1, 2], [3, 4])
nested[0].append(99) 
# Modifies the first list inside the nested tuple, demonstrating that while tuples are immutable, 
# they can contain mutable objects like lists.
print(nested)
```

## 8. Hashability

A tuple is hashable only when its elements are hashable. Hashable tuples can be used as dictionary keys or set elements.

```python
# Dictionaries require their keys to be hashable, which means they must be immutable. 
# Tuples are immutable and can be used as dictionary keys, while lists are mutable and cannot be used as dictionary keys.
d = {(1, 2): "point"}
print(d)

try:
    d = {([1, 2]): "invalid"}
except TypeError as e:
    print(type(e).__name__, e) # TypeError: unhashable type: 'list'
# Set elements must be hashable, and lists are mutable and therefore not hashable. 
# Tuples, being immutable, can be used as set elements or dictionary keys.
s=(set([1, 2, 3]), set([4, 5]))
print(s)
```

## 9. Iteration

Tuples are iterable and can be traversed directly.

```python
t = ("a", "b", "c")

for item in t:
    print(item)

for index, item in enumerate(t): # Enumerate provides both the index and the item from the tuple t during iteration.
    print(index, item)
```

## 10. Complexity Quick Reference

| Operation | Complexity |
|---|---:|
| Indexing | O(1) |
| `len()` | O(1) |
| Membership | O(n) |
| `count()` | O(n) |
| `index()` | O(n) |
| Concatenation | O(n + m) |
| Repetition | O(nk) |
| Slicing | O(k) |
| Iteration | O(n) |

Space for a tuple is O(n) for the tuple's element references. A slice or concatenation creates a new tuple and therefore requires additional space proportional to the resulting tuple.

These are typical operational costs; do not treat CPython implementation details as language-level guarantees.

## 11. Mutability and Memory Notes

- Tuples are immutable sequences.
- A tuple stores references to its elements.
- `a = b` aliases the same tuple.
- A tuple does not recursively make its contained objects immutable.
- Tuple slicing creates a new tuple.
- Tuple immutability makes tuples useful for fixed collections and, when all elements are hashable, as dictionary keys/set elements.

```python
a = (1, 2, 3)
b = a
c = a[:]

print(a is b)
print(a is c)
print(a == c)
```
