# Strings

## 1. String fundamentals

A Python string is an object of type `str` representing text. Strings are sequences, so they support indexing, slicing, iteration, membership testing, and `len()`.

String literals can use single quotes, double quotes, or matching triple quotes.

````python
single = 'hello'
double = "hello"
triple = '''hello
world'''

print(single, type(single))
print(double, type(double))
print(triple, type(triple))

text = "Python"
print(len(text))
print(text[0])
print(text[-1])
print(text[1:4])
print("Py" in text)
````

### String indexing and slicing

Indexes start at `0`. Negative indexes count from the end.

Slicing uses:

```python
s[start:stop:step]
```

`start` is included and `stop` is excluded. A slice creates another string value.

````python
text = "Python"

print(text[0])
print(text[-1])
print(text[:4])
print(text[2:5])
print(text[::2])
print(text[::-1])
````

## 2. Strings are immutable

An existing string cannot be changed in place. Operations that appear to modify a string instead produce a new string.

````python
text = "Python"

try:
    text[0] = "J"
except TypeError as error:
    print(type(error).__name__, error)

text = "J" + text[1:]
print(text)
````

````python
text = "python"
new_text = text.upper()

print("original:", text)
print("new:", new_text)
````

## 3. Iterating over strings

The most direct approach is:

```python
for character in text:
    ...
```

Use `enumerate()` when both position and character are needed.

````python
text = "Python"

for character in text:
    print(character)
````

````python
for index, character in enumerate("Python"):
    print(index, character)
````

### Iterator protocol

`iter(string)` creates an iterator and `next(iterator)` retrieves successive characters. A `for` loop handles this machinery automatically and stops when `StopIteration` occurs.

````python
text = "cat"
iterator = iter(text)

print(next(iterator))
print(next(iterator))
print(next(iterator))

try:
    print(next(iterator))
except StopIteration:
    print("iterator exhausted")
````

## 4. String comparisons and membership

Strings support equality and ordering comparisons. Membership uses `in` and `not in` to test for a substring.

````python
print("apple" == "apple")
print("apple" == "Apple")
print("apple" < "banana") # lexicographical comparison

text = "Python programming"
print("Python" in text)
print("gram" in text)
print("Java" not in text)
````

## 6. Case conversion

````python
text = "hello Python WORLD"

print(text.lower()) # converts to lowercase
print(text.upper()) # converts to uppercase
print(text.casefold()) # converts to lowercase (more aggressive) 
# difference between lower() and casefold() is that casefold() is more aggressive in converting characters to lowercase, especially for certain Unicode characters. 
# For example, the German letter 'ß' will be converted to 'ss' using casefold(), while lower() will leave it unchanged. 
print(text.capitalize()) # converts first character to uppercase and the rest to lowercase
print(text.title()) # converts first character of each word to uppercase and the rest to lowercase, camel case
print(text.swapcase()) # swaps the case of each character
````

`casefold()` is intended for caseless matching and can be more aggressive than `lower()` for Unicode text.

````python
a = "Straße"
b = "STRASSE"

print(a.lower() == b.lower())
print(a.casefold() == b.casefold())
````

## 7. Searching: `find`, `index`, `rfind`, `rindex`, `count`

- `find()` returns the first matching index or `-1`.
- `index()` returns the first matching index or raises `ValueError`.
- `rfind()` / `rindex()` search from the right.
- `count()` counts non-overlapping occurrences.

````python
text = "banana"

print(text.find("ana")) # returns the indec of the first occurrence of the substring, or -1 if not found
print(text.find("xyz")) # returns -1 if the substring is not found
print(text.index("ana")) # returns the index of the first occurrence of the substring, or raises a ValueError if not found

try:
    print(text.index("xyz")) # raises a ValueError if the substring is not found
except ValueError as error:
    print(type(error).__name__, error) 

print(text.rfind("ana")) # returns the index of the last occurrence of the substring, or -1 if not found
print(text.rindex("ana")) # returns the index of the last occurrence of the substring, or raises a ValueError if not found
print(text.count("ana")) # returns the number of non-overlapping occurrences of the substring
````

## 8. `startswith()` and `endswith()`

Use these when the question is specifically about a prefix or suffix. They also accept tuples of alternatives.

````python
filename = "photo.jpeg"

print(filename.startswith("photo"))
print(filename.endswith(".jpeg"))
print(filename.endswith((".jpg", ".jpeg", ".png")))
print(filename.startswith(("img_", "photo_", "pic_")))
````

## 9. Splitting: `split`, `rsplit`, `splitlines`

`split()` returns a list of substrings. `rsplit()` splits from the right. `splitlines()` splits text into lines.

````python
text = "Python is easy to learn"

print(text.split()) # splits the string into a list of words using whitespace as the delimiter
print(text.split(" ")) # splits the string into a list of words using space as the delimiter
print(text.split(" ", 2)) # splits the string into a list of words using space as the delimiter, but only splits at the first 2 occurrences

path = "home/user/projects/python/file.py"
print(path.split("/", 2)) 
print(path.rsplit("/", 2)) # splits the string into a list of words using slash as the delimiter, but only splits at the last 2 occurrences

lines = "first\nsecond\nthird"
print(lines.splitlines() )# splits the string into a list of lines using the newline character as the delimiter
````

## 10. `partition()` and `rpartition()`

These return a 3-tuple:

```text
(before, separator, after)
```

`partition()` uses the first separator occurrence; `rpartition()` uses the last.

````python
email = "user@example.com"

print(email.partition("@"))
print(email.rpartition("@"))
print("no-separator".partition("@"))
````

## 11. `join()`

`separator.join(iterable)` combines strings with the separator between elements.

For many pieces, `join()` is the standard efficient string-building pattern.

````python
words = ["Python", "is", "fun"]

print(" ".join(words))
print("-".join(["2026", "09", "16"]))
print("".join(["a", "b", "c"]))
````

````python
numbers = [1, 2, 3]

result = ", ".join(str(number) for number in numbers)
print(result)
````

## 12. `replace()`

`replace(old, new, count)` returns a new string. `count` can limit replacements.

````python
text = "one two two three"

print(text.replace("two", "TWO"))
print(text.replace("two", "TWO", 1))
print("original:", text)
````

## 13. `strip()`, `lstrip()`, `rstrip()`

By default these remove whitespace at the corresponding ends. With an argument, the argument is treated as a **set of characters**, not as one exact substring.

````python
text = "   Python   "

print(repr(text.strip()))
print(repr(text.lstrip()))
print(repr(text.rstrip()))

text = "---Python---"
print(text.strip("-"))
````

## 14. Character classification

Common Boolean methods include:

`isalpha`, `isalnum`, `isascii`, `isdecimal`, `isdigit`, `isnumeric`, `isspace`, `islower`, `isupper`, `istitle`, `isidentifier`, `isprintable`.

````python
samples = ["Python", "Python3", "123", "  ", "Hello World", "variable_name"]

for value in samples:
    print(
        repr(value),
        "alpha=", value.isalpha(),
        "alnum=", value.isalnum(),
        "space=", value.isspace(),
        "lower=", value.islower(),
        "upper=", value.isupper(),
        "title=", value.istitle(),
        "identifier=", value.isidentifier(),
    )
````

### `isdecimal()`, `isdigit()`, `isnumeric()`

These are related but not identical Unicode classifications. ASCII `"123"` returns `True` for all three; other Unicode numeric characters can differ.

````python
for value in ["123", "²", "Ⅷ"]:
    print(
        repr(value),
        "decimal=", value.isdecimal(),
        "digit=", value.isdigit(),
        "numeric=", value.isnumeric(),
    )
````

## 15. Alignment and padding

````python
text = "Python"

print(text.center(12, "-"))
print(text.ljust(12, "."))
print(text.rjust(12, "."))
print("42".zfill(5))
````

## 16. `translate()` and `maketrans()`

`str.translate()` can apply a character translation table. `str.maketrans()` helps create that table.

````python
table = str.maketrans({
    "a": "@",
    "e": "3",
    "i": "1",
    "o": "0",
})

print("safe code".translate(table))
````

# 18. Working with strings efficiently

Strings are immutable. Repeatedly constructing a growing string inside a large loop can cause repeated allocation and copying.

For many pieces, prefer:

```python
parts = []
for piece in pieces:
    parts.append(piece)

result = "".join(parts)
```

For a small number of values, ordinary `+` concatenation is perfectly readable.

````python
# Small concatenation: clear and appropriate
first = "Amrutha"
last = "Darshini"
full_name = first + " " + last
print(full_name)

# Many pieces: collect, then join
parts = []
for word in ["Python", "strings", "are", "immutable"]:
    parts.append(word)

print(" ".join(parts))
````

# 19. f-strings

An f-string is a formatted string literal prefixed with `f` or `F`.

Basic syntax:

```python
f"text {expression} text"
```

Expressions inside replacement fields are evaluated at runtime.

Official documentation:
https://docs.python.org/3.14/reference/lexical_analysis.html#f-strings


````python
name = "Amrutha"
age = 21

message = f"My name is {name} and I am {age} years old."
print(message)
````

## 20. f-string expressions

Replacement fields can contain Python expressions, including method calls and calculations.

````python
name = "amrutha"
price = 100
quantity = 3

print(f"{name.title()}")
print(f"Total: {price * quantity}")
print(f"Uppercase: {'python'.upper()}")
````

## 21. f-string formatting

A format specification follows `:`.

Common examples:

- `.2f` : two decimal places
- `,` : thousands separator
- `>`, `<`, `^` : alignment
- `08d` : zero-padded integer
- `%` : percentage


````python
value = 1234567.89123

print(f"{value:.2f}") # formats the value to 2 decimal places
print(f"{value:,}") # formats the value with commas as thousands separators
print(f"{value:,.2f}") # formats the value with commas as thousands separators and 2 decimal places
print(f"{value:>15,.2f}") # right-aligns the formatted value in a field of width 15

count = 42
print(f"{count:08d}") # zero-pads the integer to a width of 8 characters
print(f"{count:^10}") # centers the integer in a field of width 10 characters

ratio = 0.875
print(f"{ratio:.2%}") # formats the value as a percentage with 2 decimal places
````

## 22. f-string conversions

The conversion flags are:

- `!s` → `str()`
- `!r` → `repr()`
- `!a` → `ascii()`


````python
text = "hello\nworld"

print(f"{text!s}")
print(f"{text!r}")
print(f"{text!a}")
````

# 26. Raw strings: `r"..."`

A raw string literal is prefixed with `r` or `R`.

Its backslashes are treated as literal characters rather than having their normal escape-sequence meaning.

Typical uses:
- regular-expression patterns
- Windows-style paths
- other backslash-heavy text

Official documentation:
https://docs.python.org/3.14/reference/lexical_analysis.html#raw-string-literals


````python
normal = "C:\\Users\\Amrutha\\Documents"
raw = r"C:\Users\Amrutha\Documents"

print(normal)
print(raw)
print(normal == raw)
````

## 27. Normal string vs raw string

Compare:

```python
"\n"
```

with:

```python
r"\n"
```

The first represents a newline. The second represents two literal characters: backslash and `n`.

````python
normal = "\n"
raw = r"\n"

print(repr(normal)) # prints the string representation of the normal string, which is a newline character
print(repr(raw)) # prints the string representation of the raw string, which is the literal characters "\n"
print(len(normal)) # prints the length of the normal string, which is 1
print(len(raw)) # prints the length of the raw string, which is 2
````

### Raw strings are not completely escape-free

A raw string cannot end with a single backslash because the final backslash would escape the closing quote.

So this is invalid:

```python
r"C:\"
```

A raw string can contain a backslash before a quote, but the backslash remains in the resulting value.

````python
# Valid: ends with two backslashes
path = r"C:\\" 
print(repr(path))
````

# 28. Raw strings and regular expressions

Raw strings are especially readable for regular expressions because regex syntax commonly contains backslashes.

````python
import re

pattern = r"\d+" # matches one or more digits
text = "Order 123, item 45"

print(re.findall(pattern, text))
````

# 32. String memory and immutability

Transformations such as `upper()`, `replace()`, and many slices produce new string values. The original string is unchanged.


````python
text = "python"
same_value = text
new_value = text.upper()

print(text, id(text))
print(same_value, id(same_value))
print(new_value, id(new_value))

print(text == same_value)
print(text is same_value)
print(text == new_value)
````

# 33. Complexity and memory

Let `n` be the length of a string and `k` the length of a produced slice.

| Operation | Typical time model | Result/additional space |
|---|---:|---:|
| `len(s)` | O(1) | O(1) |
| `s[i]` | O(1) | O(1) for the resulting value |
| `s[a:b]` | O(k) | O(k) |
| Iterating through `s` | O(n) | O(1) iterator overhead |
| `"x" in s` | O(n) worst case | O(1) auxiliary |
| `s.replace(...)` | O(n) scale | O(n) result |
| `s.split(...)` | O(n) scale | O(n) result storage |
| `sep.join(parts)` | O(total output size) | O(total output size) result |
| `s.upper()` | O(n) scale | O(n) result |
| `s.strip()` | O(n) worst case | up to O(n) result |

Exact string-search and memory behavior can depend on the implementation and input. Treat these as practical complexity models.

**Key point:** `len()` does not scan the string; strings maintain their length. But operations that construct transformed strings generally need to process/copy the relevant characters.


# 34. Integrated example

This combines iteration, `strip()`, `split()`, `enumerate()`, `join()`, f-strings, and formatting.

````python
raw_lines = [
    " Alice, 92 ",
    "Bob,85",
    " Charlie, 78 ",
]

records = []

for line in raw_lines:
    name, score = line.strip().split(",")
    records.append((name.strip(), int(score)))

for position, (name, score) in enumerate(records, start=1):
    print(f"{position}. {name:<10} {score:>3}")
````

### Complexity of the integrated example

If there are `n` lines and `N` total input characters:

- Parsing is approximately O(N).
- `records` uses O(n) container space plus storage for the parsed strings.
- The output loop is O(n) plus the amount of text produced.
- `enumerate()` itself does not materialize the whole sequence.
