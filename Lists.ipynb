# Python Lists

**Official documentation:**  
- [Python Built-in Types — Lists](https://docs.python.org/3/library/stdtypes.html#lists)  
- [Python Tutorial — More on Lists](https://docs.python.org/3/tutorial/datastructures.html#more-on-lists)  
- [Python Time Complexity](https://wiki.python.org/moin/TimeComplexity)

**Core idea:** A list is a mutable sequence. It preserves order, supports duplicate values, indexing and slicing, and can contain arbitrary Python objects.

## 1. Creating Lists

`[]` creates an empty list. `list(iterable)` creates a list containing the iterable's items in order.

````python
empty = []
numbers = [10, 20, 30]
from_tuple = list((1, 2, 3))
from_string = list("abc")
from_range = list(range(5))

print(empty)
print(numbers)
print(from_tuple)
print(from_string)
print(from_range)
````

## 2. Indexing and Slicing

Indexing is zero-based. Negative indexes count from the end. Slicing creates a new list.

````python
a = [10, 20, 30, 40, 50]

print(a[0])
print(a[-1])
print(a[1:4])
print(a[:3])
print(a[::2])
print(a[::-1])
````

## 3. Common Sequence Operations

Lists support the common sequence operations documented by Python: membership, concatenation, repetition, indexing, slicing, `len()`, and comparisons.

````python
a = [1, 2, 3]
b = [4, 5]

print(2 in a)
print(9 not in a)
print(a + b)
print(a * 2)
print(a[1])
print(a[1:])
print(len(a))
print([1, 2] == [1, 2])
print([1, 2] < [1, 3])
````

## 4. Mutable Sequence Operations

Lists can be changed in place using indexed assignment, deletion, slice assignment, `+=`, and `*=`.

````python
a = [1, 2, 3, 4]

a[1] = 20
print(a)

del a[2]
print(a)

a[1:2] = [7, 8]
print(a)

a += [9, 10]
print(a)

a *= 2
print(a)
````

## 5. `append()`

Adds one object to the end of the list. It mutates the list and returns `None`.

````python
a = [1, 2]
result = a.append(3)

print(a)
print(result)
````

## 6. `extend()`

Appends every item from an iterable. It mutates the list and returns `None`.

````python
a = [1, 2]
result = a.extend([3, 4])

print(a)
print(result)
````

## 7. `insert()`

Inserts an item before the specified position. It mutates the list and returns `None`.

````python
a = [1, 3]
result = a.insert(1, 2)

print(a)
print(result)
````

## 8. `remove()`

Removes the first item equal to the given value. Raises `ValueError` if the value is absent.

````python
a = [1, 2, 2, 3]
a.remove(2)
print(a)

# a.remove(99)  # ValueError
````

## 9. `pop()`

Removes and returns an item. By default it removes the last item; an index can be supplied.

````python
a = [10, 20, 30]

last = a.pop()
first = a.pop(0)

print(last)
print(first)
print(a)
````

## 10. `clear()`

Removes all items from the list. Returns `None`.

````python
a = [1, 2, 3]
result = a.clear()

print(a)
print(result)
````

## 11. `index()`

Returns the zero-based index of the first matching value. `start` and `stop` can limit the search. Raises `ValueError` if not found.

````python
a = ["a", "b", "c", "b"]

print(a.index("b"))
print(a.index("b", 2))
# print(a.index("x"))  # ValueError
````

## 12. `count()`

Returns the number of times a value occurs in the list.

````python
a = [1, 2, 2, 2, 3]
print(a.count(2))
print(a.count(9))
````

## 13. `reverse()`

Reverses the list in place. Returns `None`.

````python
a = [1, 2, 3]
result = a.reverse()

print(a)
print(result)
````

## 14. `sort()`

Sorts the list in place. `key` selects a comparison key and `reverse=True` reverses the resulting order. The sort is stable. `sort()` returns `None`.

````python
a = [3, 1, 2]
result = a.sort()

print(a)
print(result)

words = ["Banana", "apple", "Cherry"]
words.sort(key=str.lower)
print(words)

numbers = [1, 2, 3]
numbers.sort(reverse=True)
print(numbers)
````

## 15. `copy()`

Returns a shallow copy. The outer list is a different list, but nested objects are still shared.

````python
a = [1, 2, 3]
b = a.copy()

print(a is b)
print(a == b)

nested = [[1, 2], [3, 4]]
copy_nested = nested.copy()
copy_nested[0].append(99)

print(nested)
print(copy_nested)
````

## 16. List Comprehensions

A list comprehension creates a new list from an iterable, optionally applying a condition.

````python
squares = [x * x for x in range(5)]
even_squares = [x * x for x in range(10) if x % 2 == 0]

print(squares)
print(even_squares)
````

## 17. Iteration

Lists are iterable. `enumerate()` is useful when both the index and value are needed.

````python
items = ["a", "b", "c"]

for item in items:
    print(item)

for index, item in enumerate(items):
    print(index, item)
````

## 18. `list()` and `sorted()`

`list(iterable)` creates a list. `sorted(iterable)` returns a new sorted list without modifying the original iterable.

````python
values = (3, 1, 2)

a = list(values)
b = sorted(values)

print(a)
print(b)

original = [3, 1, 2]
result = sorted(original)

print(original)
print(result)
````

## 19. Complexity Quick Reference

For a Python list, common operation costs are:

| Operation | Complexity | Notes |
|---|---:|---|
| Index `a[i]` | O(1) | Direct positional access |
| Assignment `a[i] = x` | O(1) | Replaces one position |
| `len(a)` | O(1) | |
| `append(x)` | O(1) amortized | End insertion |
| `pop()` | O(1) | Removes last item |
| `pop(i)` | O(n) | Shifting may be required |
| `insert(i, x)` | O(n) | Elements may need shifting |
| `remove(x)` | O(n) | Searches for first match |
| `x in a` | O(n) | Linear search |
| `count(x)` | O(n) | Scans the list |
| `index(x)` | O(n) | Linear search |
| `reverse()` | O(n) | |
| `sort()` | O(n log n) | Typical comparison sort bound |
| Slicing | O(k) | `k` = number of elements copied |
| Copy | O(n) | Shallow copy |

Complexity can depend on the operation and implementation. Treat CPython-specific details as implementation details, not language guarantees.

## 20. Space / Memory Notes

- A list is a **mutable sequence**.
- List elements are references to Python objects; the objects themselves are not necessarily stored inside the list.
- `a = b` creates another reference to the same list.
- `a.copy()` creates a new outer list (**shallow copy**).
- A slice such as `a[:]` also creates a new list.
- List comprehensions and slices require additional space for the newly created list.
- Dynamic resizing is an implementation concern; do not rely on a particular capacity or allocation strategy.

````python
a = [1, 2, 3]
b = a
c = a.copy()

print(a is b)  # same list
print(a is c)  # different outer list
````

## 21. Mutability

A list can be changed after creation.

```python
a = [1, 2]
a.append(3)
```

This changes the existing list rather than creating an immutable sequence.

````python
a = [1, 2]
print(a)

a.append(3)
print(a)
````

## 22. Important Pitfalls

- `append([3, 4])` adds **one list object**; `extend([3, 4])` adds two elements.
- Methods such as `append()`, `extend()`, `insert()`, `remove()`, `reverse()`, and `sort()` modify the list and return `None`.
- `remove()` removes by value; `pop()` removes by index and returns the removed item.
- `sort()` mutates the list; `sorted()` returns a new sorted list.
- `copy()` is shallow, not recursive.
- Modifying a list while iterating over it can produce unexpected results.

````python
a = [1, 2]
a.append([3, 4])
print(a)

b = [1, 2]
b.extend([3, 4])
print(b)
````

## 23. Quick Revision

**List = ordered + mutable + indexable + sliceable + duplicates allowed**

Use a list when you need a mutable ordered collection and positional access.
